### 3.1 Autograd: Tự động tính đạo hàm.
Trung tâm của toàn bộ các mạng nơ ron hoạt đọng trên pytorch là autograd package.

`Chức năng của autograd`: Tự động tính toán đoaọ hàm trên toàn bộ các toán tử của tensor. Nó là một framework được định nghĩa trong quá trình chạy, có nghĩa rằng quá trình lan  truyền ngược được xác định khi mà code được chạy. và do đó mỗi vogngf lặp có thể có kết quả thay đổi tham số theo lan truyền ngược khác nhau.

`Theo dõi lịch sử của tensor torch`: torch.tensor là package khởi tạo các tensor torch. Mỗi một tensor torch sẽ có 1 thuộc tính là `.requires_grad`, nếu ta sét thuộc tính này về True, các toán tử triển khai trên tensor sẽ được theo dõi. Kgi kết thuốc quá trình lan truyền thuận (hoặc quá trình tính toán output) ta có thể gọi `.backward()` và mọi tính toán gradinent sẽ được tự động thực hiện dựa trên lịch sử đã được lưu lại. các gradient cho tensor này sẽ được tích lũy và xem lại thuộc tính.

Để dừng theo dõi một tensor chúng ta gọi hàm `.detach()` khi đó các hoạt động trên tensor sẽ không còn được lưu vết nữa.

Ngoài ra để ngăn tensor lưu lại lich sử (và sử dụng memory), chúng ta có thể bao quanh code block triển khai tensor với hàm `with torch.no_grad():` nó rất hữu ích trong trường hiowj đánh giá model bởi vì khi thuộc tính `requires_grad = True` thì model sẽ có thể được cập nhật tham số. Nhưng quá trình đánh giá model sẽ không cần cập nhật tham số nên chúng ta cần áp dụng gradient lên chúng. Đơn giản là set `requires_grad = False`.

**Ví dụ về lưu trữ đồ thị tính toán**
Bên dưới ta sẽ khởi tạo một tensor torch có khả năng thay đổi theo 2 cách
- Cách 1: set `requires_grad = True`

```python
import torch
x = torch.ones(2,2,requires_grad=True)
print(x)
y = x+2
print(y)
```

Do tại x ta thoeo dõi thay đổi bằng các set tham số `requires_grad=True` nên các tính toán được thực hiện trên x sẽ được theo dõi lại ở các thuộc tính `grad_fn`

```python
print(y.grad_fn)
```

giá trị của `grad_fn` cho thấy ta đã thực hiện một phép cộng để thu được y. thực hiện tiếp 1 biến đổi nữa sử dụng $y$

```python
z = y*y*3
out = z.mean()
print(z)
print(out)
```

- Cách 2: theo cách inplace. Mặc định của `requires_grad` khi khởi tạo 1 torch là False tức là sẽ không ghi lại lịch sử thay đổi

```python
a = torch.ones(3, 3)
print(a)
print(a.requires_grad)
a.requires_grad_(True)
print(a.requires_grad)
b = (a*a).sum()
print(b)
print(b.grad_fn)
```

Kết quả của gradient $d(out)/dx$ sẽ được lưu trong phần từ x.grad

Chúng ta nhận được ma traanjj chỉ gồm các phần tử là 4.5. Đây chính là đạo hàm của mỗi phần tử của x theo $y.\frac{dy}{dx}$

Hàm `torch.autograd` sẽ là hàm chức năng tính tích giữa vector và ma trận jacobian. Hàm số cho ta biết mức độ thay đổi của các chiều khi đi theo phương gradient.

```python
import torch

x = torch.randn(3, requires_grad = True)
yhat = torch.randn(3, requires_grad = True)*2
y = x*2
l = ((y-yhat)**2).mean()
print(l)
```

trong TH này $y$ sẽ không còn là 1 scalar. Hàm `torch.autograd` sẽ không tính toán ma trận jacobian trực tiếp mà thay vào đó sẽ tính tích vector-jacobian theo vector $v$ truyền vào đối số `.backward`

```python
# Khởi tạo một vector gradient tự do v
v = torch.tensor([0.1, 1.0, 0.001], dtype = torch.float)
# Tính ma trận Jacobian (đạo hàm của y theo v)
y.backward(v)
# Tính tích vector-jacobian chính là đạo hàm của 
print(x.grad)
```

Để dừng autograd theo dõi các thay đổi lịch sử trên tensor, chúng ta có thể thiết lập `.requires_grad=True` hoặc đặt các biến đổi tensor trong block code `torch.no_grad()`.

```python
print(x.requires_grad)
print((x*x).requires_grad)
with torch.no_grad():
  print((x*x).requires_grad)
```

## Thực hành trên mạng neural network kiến trúc cnn.
Các mạng neural sẽ được xây dựng dựa trên package torch.nn. Dựa trên autograd model sẽ xác định đạo hàm bậc 1 theo các chiều dữ liệu. Một nn.Model sẽ gồm các layers và một phương thức `forward(input)` để trả ra kết quả `output`.

```python
import torch
import torch.nn as nn
import torch.nn.functional as F

class Net(nn.Module):
    def __init__(self):
        super(Net,self).__init__()
        self.conv1 = nn.Conv2d(1,6,3)
        self.conv2 = nn.Conv2d(6,16,3)
        self.fc1 = nn.Linear(16*5*5,120)
        self.fc2 = nn.Linear(120, 84)
        self.fc3 = nn.Linear(84, 10)
    
    def forward(self, x):
        x = F.max_pool2d(F.relu(self.conv1(x)), (2,2))
        x = F.max_pool2d(F.relu(self.conv2(x)), 2)
        x = x.view(-1, self.num_flat_features(x))
        x = F.relu(self.fc1(x))
        x = F.relu(self.fc2(x))
        x = self.fc3(x)
        return x
    
    def num_flat_features(self, x):
        size = x.size()[1:]
        num_features = 1
        for s in size:
            num_features *= s
        return num_features
```

```python
net = Net()
print(net)
```

chúng ta phải xác định trước `forward` function để trả ra kết quả của model ở đầu ra. Dựa trên `forward` function, hàm `backward` function (là nơi mà gradient tại mỗi layers được tính toán) sẽ được tự đọng sác định khi ta sử dụng `autograd`. Chúng ta cũng có thể sử dụng bất kì một phép biến đổi toán tử Tensor nào trên hàm forward function.

Các tham số huấn luyện (tham số mà có thể thay đổi được trong huấn luyện) của mô hình được  trả về bằng hàm `net.parameters()`

```python
params = list(net.parameters())
print(len(params))
print(f"conv1's weight: {params[0].size()}")
```

Chúng ta thử nghiệm khởi tạo một đầu vào ngẫu nhiên 32x32. Chú ý rằng: kì vọng đầu vào của mạng lenet là 32x32.

```python
input = torch.randn(1, 1, 28, 28)
out = net(input)
print(out)
```

khi đó kết quả đầy ra thu được là một tensor có 10 phần tử. mỗi phần tử tương ứng với điểm số được phân bố trong class mà nó thuộc về. Bên dưới chúng ta chuyển toàn bộ các gradients trong bộ nhớ đệm và 0 bằng hàm zero_grad() và lan truyền ngược với gradient ngẫu nhiên.

```python
net.zero_grad()
out.backward(torch.randn(1,10))
```

**Lưu ý:** `torch.nn` chỉ hỗ trợ các mini-batches. Toàn bộ `torch.nn` packages chỉ hỗ trợ đầu vaofd là mini-batch của mẫu (tức là luôn có 1 chiều trong shape qui định batch size), và không tiếp nhận một mẫu riêng lẽ.

Chẳng hạn, `nn.Conv2d` sẽ nhận đầu vào là 4D Tensor cuẩ nSamples xs nChannel x Height x Width. Trong đó chiều đầu tiên là kích thước mẫu ( batch size).

Nếu bạn có một mẫu đơn lẽ, chỉ cần sử dụng input.unsqueeze(0) để thêm vào một chiều `batch size` giả mạo.

Tổng kết:
- torch.Tensor: là một mảng nhiều chiều hỗ trợ các biến đổi autograd như `backward()`. và cũng lưu trữ các gradients của tensor.
- nn.Module: Neural netword module. Thuận tiện trong gói các tham số với sự hỗ trợ để đẩy chúng lên GPU, export và loading tham số, ...
- autograd.Function: kế thứa quá lan truyền thuận và lan truyền ngược của một biến đổi autograd. Mọi triển khai Tensor tạo ra ít function node kết nối đến function được tạo bởi tensor và mã hóa lịch sử của chúng.

### 1. Hàm Loss
Một hàm loss sẽ nhận 1 cặp (output, target) và tính toán giá trị khoảng cách giữa output và giá trị target.

Có một số dạng loss function khác nhau mà ta có thể tham khảo được hỗ trợ trong nn package: [Loss function trong nn](https://pytorch.org/docs/stable/nn.html#transformer-layers). Dạng đơn giản nhất là `nn.MSELoss` (trung bình bình phương sai số) tính toán trung bình bình phương sai số giữa giá trị output và giá trị target.

```python
output = net(input)
target = torch.randn(10)
target = target.view(1,-1)
criterion = nn.MSELoss()
loss = criterion(output, target)
print(loss)
```

Tiến trình backward của hàm loss function sẽ sử dụng thuộc tính `.grad_fn` của nó để lần tìm trên đồ thị tính toán quá trình biến đổi tensor như bên dưới:

`input -> conv2d -> relu -> maxpool2d -> conv2d -> relu -> maxpool2d -> view -> linear -> relu -> linear -> relu -> linear -> MSELoss -> loss`

Khi ta gọi và hàm `loss.backward()` toàn bộ graph sẽ tính toán đạo hàm của loss function, các tensors trong graph có thuộc tính `requires_grad = True` thì sẽ có tensor .grad được cập nhật gradient theo trình tự lũy tiến

```python
print(loss.grad_fn) # MSELoss
print(loss.grad_fn.next_functions[0][0]) #Linear
print(loss.grad_fn.next_functions[0][0].next_functions[0][0]) # Relu
```

### 2. Lan truyền ngược (backpropagation)
Để lan truyền ngược sử dụng hàm `loss.backward()`. Nhưng trước đó ta cần xóa những gradient đang có và các grandient khác sẽ tích lũy vào gradient hiện có.

```python
net.zero_grad() # chuyển vễ 0 toàn bộ các gradient trong bộ nhớ đẹm của toàn bộ các parameters.
print('conv1.bias.grad before backward')
print(net.conv1.bias.grad)
loss.backward()
print('conv1.bias.grad after backward')
print(net.conv1.bias.grad)
```

### 3 cập nhật trọng số.
công thức đơn giản để cập nhật trọng số là:

Chúng ta có thể triển khai bằng sử dụng python code đơn giản sau:

```python
learning_rate = 0.01
for f in net.parameters():
    f.data.sub_(f.grad.data * learning_rate)
    
```

trong đó `sub_` là một hàm inplace của phép trừ.
Tuy nhiên khi sử dụng mạng neural networks, ta muốn sử dụng đa dạng các phương pháp cập nhật gradient khác nhau như: SGD, Nesterov-SGD, Adam, RMSProp, ... Do đó sử dụng package `torch.optim` chúng ta có thể thực hiện được toàn bộ những phương pháp gradient descent này một cách đơn giản.

```python
import torch.optim as optim

optimizer = optim.SGD(net.parameters(), lr=0.001, momentum=0.9)
# in your optimizer
optimizer.zero_grad() # Zero gradient buffers
output = net(input)
loss = criterion(output, target)
loss.backward()
optimizer.step() # thực hiện update.
```

Hàm `criterition()` được sử dụng để tính loss function. `loss.backward()` sẽ thực thiện quá trình lan truyền ngược và `optimizer.step()`.
